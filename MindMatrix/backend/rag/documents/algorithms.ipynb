## Two Sum

Given an array of integers and a target value, find two
indices whose values add up to the target.

The optimal solution uses a HashMap.

Time Complexity: O(n)

Space Complexity: O(n)

```python
def two_sum(nums, target):
    """
    Find two indices whose values add up to target.

    Time Complexity: O(n)
    Space Complexity: O(n)
    """

    # Dictionary stores previously visited numbers
    seen = {}

    # Traverse the array
    for i, num in enumerate(nums):

        # Calculate the required complement
        complement = target - num

        # Check whether the complement was already seen
        if complement in seen:
            return [seen[complement], i]

        # Store the current number and its index
        seen[num] = i

    # Return empty list if no pair exists
    return []
```

## Binary Search

Binary Search works on a sorted array.

Instead of checking every element, it repeatedly divides
the search space into two halves.

Time Complexity: O(log n)

Space Complexity: O(1)

```python
def binary_search(nums, target):
    """
    Search for target in a sorted array.

    Time Complexity: O(log n)
    Space Complexity: O(1)
    """

    # Initialize the search boundaries
    left = 0
    right = len(nums) - 1

    # Continue while the search range is valid
    while left <= right:

        # Find the middle index
        mid = (left + right) // 2

        # Target found
        if nums[mid] == target:
            return mid

        # Target is in the right half
        elif nums[mid] < target:
            left = mid + 1

        # Target is in the left half
        else:
            right = mid - 1

    # Target not found
    return -1
```

## Sliding Window

Sliding Window is useful for problems involving
contiguous subarrays or substrings.

Common applications:
- Maximum sum subarray
- Longest substring
- Minimum window substring

Typical Time Complexity: O(n)

```python
def max_sum_subarray(nums, k):
    """
    Find the maximum sum of a subarray of size k.

    Time Complexity: O(n)
    Space Complexity: O(1)
    """

    # Check whether a valid window is possible
    if len(nums) < k:
        return None

    # Calculate the first window
    window_sum = sum(nums[:k])

    # Store the maximum sum found
    maximum = window_sum

    # Move the window through the array
    for right in range(k, len(nums)):

        # Add the new element
        window_sum += nums[right]

        # Remove the element leaving the window
        window_sum -= nums[right - k]

        # Update maximum
        maximum = max(maximum, window_sum)

    return maximum
```
